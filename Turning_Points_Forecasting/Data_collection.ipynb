```python
import os
import datetime
import yfinance as yf
import pandas as pd

# data interval
interval = "5m" #1m, 2m, 5m, 15m, 60m, 90m, 1h, 1d, 5d, 1wk, 1mo, 3mo
# period ='1mo'

# limits
# 1m --> 8 days
# 2m, 5m, 15m, 90m  --> 60 days
# 60m, 1h, --> 730 days
# 1d, 5d, 1wk, 1mo, 3mo --> no limit

# start & end date
end_date = datetime.date.today()
print(end_date)
start_date = end_date - datetime.timedelta(days=59)


# Raw data root
raw_data_root = os.getcwd()
save_path = os.path.join(raw_data_root, "OHVLC_5min", end_date.strftime("%Y-%m-%d"))
print(save_path)

if not os.path.exists(save_path):
    os.makedirs(save_path)

#  symbols
# Tickers of interest
symbols = [
    "AAPL", "AMZN", "BABA", "BHAT", "GOOGL", "INTC", "META",
    "MSFT", "NFLX", "NVDA", "PLTR", "SPY", "SHOP", "TSLA", "QQQ"
]
# symbols = ["NFLX", "META", "AMZN", "AAPL", "GOOG", "MSFT"]
```

```python
# collect data
for symbol in symbols:
  data = yf.download(symbol, start=start_date, end=end_date, interval=interval)
  # data.index = data.index.tz_convert(None)

  # Save the DataFrame to the specified path
  df = pd.DataFrame(data)
  fileName = symbol + '_OHVLC_' + interval + '.csv'
  raw_file_path =  os.path.join(save_path, fileName)
  df.to_csv(raw_file_path)
```
